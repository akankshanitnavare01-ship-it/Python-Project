# HOTEL BILLING MANAGEMENT SYSTEM

```python
print("* WELCOME TO HAPPY RESTAURANT *")
menu={
    "Pizza":160,
    "Pasta":120,
    "Burger":80,
    "Salad":70,
    "Momos":90,
    "Sandwich":50,
    "Cheese_Vadapav":35,
    "Vadapav":25,
    "Coffee":100,
    "Tea":20
}
menu 
print("Pizza : Rs.160. \nPasta : Rs.120. \nBurger : Rs.80. \nSalad : Rs.70. \nMomos : Rs.90. \nSandwich : Rs.50. \nCheese_Vadapav : Rs.35. \nVadapav : Rs.25. \nCoffee : Rs.100. \nTea : Rs.20")
```

```python
Total_bill = 0
Order = list(map(str,input("Enter Your Orders = ").split()))
for items in Order:
    if items in menu:
        price=menu[items]
        Total_bill = Total_bill+price

    else:
        print("Your order is not available yet")
print("Your order has been added")

another_order = input("Would you like to order anything else?(if yes enter 'yes' otherwise enter 'no') = ")
if another_order == 'no':
    print(f"Your Total Bill is Rs.{Total_bill} .")

elif another_order == 'yes':
    Order_2 = list(map(str,input("Enter Your Orders").split()))
    for items_2 in Order_2:
        if items_2 in menu:
            price=menu[items_2]
            Total_bill+=price
        else:
            print("Your order is not available yet")
    print(f"Your Total Bill is Rs.{Total_bill} .")
print("Thank You....!")
```
